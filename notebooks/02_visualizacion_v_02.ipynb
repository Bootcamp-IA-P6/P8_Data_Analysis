# Pruebas con scatter_mapbox y pairplot

```python
import pandas as pd
import plotly.express as px
```

```python
df_fusion = pd.read_csv('../data/raw/fusion_airbnb.csv', sep=';')

# Obtener filas y columnas
filas, columnas = df_fusion.shape
print(f'✅ Dataset fusion_airbnb cargado correctamente')
print(f'📊 Dimensiones: {filas} filas x {columnas} columnas')
```

```python
fig = px.scatter_mapbox(
    df_fusion, lat="latitude", lon="longitude", 
    color="city", size="price", hover_name="city", 
    mapbox_style="carto-positron", zoom=1,
    height=700, width=1200
)

fig.show()
```

```python
import seaborn as sns
```

```python
# Pairplot de variables numéricas
sns.pairplot(df_fusion[["price", "minimum_nights", "number_of_reviews", "city"]], hue="city")
```
